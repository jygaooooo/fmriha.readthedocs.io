# Understanding Hyperalignment (Optional Reading)

This note collects the conceptual and mathematical background for hyperalignment: why subject-specific voxel spaces need alignment, how procrustes rotation estimates a T matrix, how common templates are constructed, and how local searchlight hyperalignment extends the same idea across the brain.

As a functional alignment method, hyperalignment fundamentally aligns an fMRI data matrix organized as time points × voxels/vertices. The alignment of this matrix can be understood from two complementary perspectives: aligning its rows (response patterns) or aligning its columns (response profiles). Because the rows and columns are two views of the same matrix, alignment from either perspective also brings the information represented by the other into alignment. Pattern alignment and profile alignment therefore provide two ways of understanding the principles of hyperalignment. Below, we present these two perspectives in turn; readers may use whichever version best suits their needs.
